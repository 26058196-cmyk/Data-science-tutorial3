## 1. Load a real dataset and visualize some of it

## 2. Evaluate if the data is good or not. Does it have enough samples?

```python
print("Rows and columns:", df.shape)
print("Duplicated rows:", df.duplicated().sum())
```

## 3. How many features does it have? Is it enough?

```python
print("All columns:", df.columns.tolist())
print("Total columns:", len(df.columns))

target = "math score"
features = df.drop(columns=[target])
print("Number of features:", len(features.columns))
print("Features:", features.columns.tolist())
```

answer:Excluding the target variable (math score), we have 7 columns covering demographic details and other test scores. While 7 columns provide sufficient context for basic pattern detection, adding extra columns like home study time or school attendance would allow for richer insights.
